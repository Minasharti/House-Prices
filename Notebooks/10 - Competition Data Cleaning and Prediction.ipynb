```python
#مدل رو روی تمام داده های ترین آموزش میدم همونطور که تو تمرین قبلی دادم تا اینجا که میخوام داده های مسابقه رو پیش بینی کنم باهاش آماده باشه
import pandas as pd
import numpy as np
train = pd.read_csv("train.csv")

#######################################################
#تمیزکاری دیتا
#پرکردن خانه های خالی از جنس استرینگ
train["PoolQC"] = train["PoolQC"].fillna("No Pool")
train["MiscFeature"] = train["MiscFeature"].fillna("None")
train["Alley"] = train["Alley"].fillna("No alley access")
train["Fence"] = train["Fence"].fillna("No Fence")
train["MasVnrType"] = train["MasVnrType"].fillna("None")
train["FireplaceQu"] = train["FireplaceQu"].fillna("No Fireplace")
train["GarageQual"] = train["GarageQual"].fillna("No Garage")
train["GarageFinish"] = train["GarageFinish"].fillna("No Garage")
train["GarageType"] = train["GarageType"].fillna("No Garage")
train["GarageCond"] = train["GarageCond"].fillna("No Garage")
train["BsmtFinType2"] = train["BsmtFinType2"].fillna("No Basement")
train["BsmtExposure"] = train["BsmtExposure"].fillna("No Basement")
train["BsmtCond"] = train["BsmtCond"].fillna("No Basement")
train["BsmtQual"] = train["BsmtQual"].fillna("No Basement")
train["BsmtFinType1"] = train["BsmtFinType1"].fillna("No Basement")

train["Electrical"] = train["Electrical"].fillna(train["Electrical"].mode()[0])

#پرکردن خانه های خالی از جنس عدد
train["LotFrontage"] = train["LotFrontage"].fillna(train["LotFrontage"].median())
train["GarageYrBlt"] = train["GarageYrBlt"].fillna(0)
train["MasVnrArea"] = train["MasVnrArea"].fillna(0)

#######################################################
#مهندسی فیچر
train["HouseAge"] = 2010 - train["YearBuilt"]

train["QualLivArea"] = train["OverallQual"] * train["GrLivArea"]

train["LotArea"] = np.log1p(train["LotArea"])

train["BsmtFinishedRatio"] = np.where (train["TotalBsmtSF"] > 0, train["BsmtFinSF1"] / train["TotalBsmtSF"], 0)#چون اگر TotalBsmtSF صفر باشد، از تقسیم بر صفر جلوگیری می‌شود.

train["QualCond"] = train["OverallQual"] * train["OverallCond"]

train["TotalBath"] = ( train["FullBath"]  + 0.5 * train["HalfBath"] + train["BsmtFullBath"]+ 0.5 * train["BsmtHalfBath"])

train["TotalSF"] = train["TotalBsmtSF"] + train["GrLivArea"]

train["RemodAge"] = 2010 - train["YearRemodAdd"]

train["HasFireplace"] = (train["Fireplaces"] > 0).astype(int)

train["HasFence"] = (train["Fence"] != "No Fence").astype(int)

train["QualFireplace"] = train["OverallQual"] * train["Fireplaces"]

#######################################################

#تعیین x,y
X = train.drop(["Id", "SalePrice"], axis=1)
y = train["SalePrice"]

#تبدیل داده استرینگ به عدد
X = pd.get_dummies(X)

#لگاریتم y
y = np.log1p(y)

#تعریف مدل
from sklearn.ensemble import RandomForestRegressor
model = RandomForestRegressor(
     n_estimators=800,
    max_depth=15,
    min_samples_split=2,
    min_samples_leaf=1,
    max_features=0.5,
    random_state=42,
    n_jobs=-1)

# آموزش روی تمام داده ها
model.fit(X, y)
```

```python
#میخوام ببینم اینا که حداقل یه خونه خالی دارن اطلاعات بیشتری بگیرم ازشون
columns = [
    "PoolQC",
    "MiscFeature",
    "Alley",
    "Fence",
    "MasVnrType",
    "FireplaceQu",
    "LotFrontage",
    "GarageYrBlt",
    "GarageCond",
    "GarageFinish",
    "GarageQual",
    "GarageType",
    "BsmtCond",
    "BsmtQual",
    "BsmtExposure",
    "BsmtFinType1",
    "BsmtFinType2",
    "MasVnrArea",
    "MSZoning",
    "BsmtHalfBath",
    "Utilities",
    "Functional",
    "BsmtFullBath",
    "BsmtFinSF1",
    "Exterior1st",
    "TotalBsmtSF",
    "BsmtUnfSF",
    "BsmtFinSF2",
    "SaleType",
    "KitchenQual",
    "GarageCars",
    "GarageArea",
    "Exterior2nd"
]

with open("data_description.txt", "r", encoding="utf-8") as f: # فایل data_description.txt را باز می‌کند. بعد میخونتش و اسمشو میذاریم f
    lines = f.readlines() #تمام خطوط فایل را میخونه و داخل یک متغیر می‌ریزد.

for column in columns:
    print("=" * 70) #این رشته "=" را ۷۰ بار تکرار کن.
    print(f"{column}\n") #\n ینی برو خط بعد

    found = False

    for i, line in enumerate(lines):
        if line.strip().startswith(column + ":"): #دنبال خطی می‌گردد که مثلاً این باشد:
            found = True

            j = i
            while j < len(lines):
                if j > i and lines[j].strip() != "" and not lines[j].startswith(" "):
                    break

                print(lines[j].rstrip())
                j += 1

            break

    if not found:
        print("ستون پیدا نشد.")

    print()
```

```python
#پرکردن خانه های خالی 
#ستون‌های متنی (String)این‌ها چون NaN در واقع معنی خاصی دارند، همان معنی را جایگزین می‌کنیم
test["PoolQC"] = test["PoolQC"].fillna("No Pool")
test["MiscFeature"] = test["MiscFeature"].fillna("None")
test["Alley"] = test["Alley"].fillna("No alley access")
test["Fence"] = test["Fence"].fillna("No Fence")
test["MasVnrType"] = test["MasVnrType"].fillna("None")
test["FireplaceQu"] = test["FireplaceQu"].fillna("No Fireplace")
test["GarageCond"] = test["GarageCond"].fillna("No Garage")
test["GarageFinish"] = test["GarageFinish"].fillna("No Garage")
test["GarageQual"] = test["GarageQual"].fillna("No Garage")
test["GarageType"] = test["GarageType"].fillna("No Garage")
test["BsmtCond"] = test["BsmtCond"].fillna("No Basement")
test["BsmtQual"] = test["BsmtQual"].fillna("No Basement")
test["BsmtExposure"] = test["BsmtExposure"].fillna("No Basement")
test["BsmtFinType1"] = test["BsmtFinType1"].fillna("No Basement")
test["BsmtFinType2"] = test["BsmtFinType2"].fillna("No Basement")

# پر کردن ستون‌های متنی با پرتکرارترین مقدار
#برای mode , median , ... # بهتره از دیتای آموزش استفاده بشه چون مدل با اون دیتا آموزش دیده
test["MSZoning"] = test["MSZoning"].fillna(train["MSZoning"].mode()[0]) #mode() ممکن است یک یا چند مقدار برگرداند، چون ممکن است چند مقدار به یک اندازه تکرار شده باشند.
#بنابراین خروجی mode() یک لیست (Series) است.
#با [0] می‌گوییم اولین مقدار پرتکرار را بردار.
test["Utilities"] = test["Utilities"].fillna(train["Utilities"].mode()[0])
test["Exterior1st"] = test["Exterior1st"].fillna(train["Exterior1st"].mode()[0])
test["Exterior2nd"] = test["Exterior2nd"].fillna(train["Exterior2nd"].mode()[0])
test["SaleType"] = test["SaleType"].fillna(train["SaleType"].mode()[0])
test["KitchenQual"] = test["KitchenQual"].fillna(train["KitchenQual"].mode()[0])
#بقیه (MSZoning، SaleType، KitchenQual، ...) هیچ معنای خاصی برای NaN ندارند، پس بهترین کار استفاده از پرتکرارترین مقدار (mode) است.

# طبق توضیح دیتاست، اگر مقدار Functional خالی بود، حالت معمولی فرض می‌شود.
test["Functional"] = test["Functional"].fillna("Typ") #Functional خود دیتاست نوشته:Assume typical unless deductions are warranted
#یعنی اگر معلوم نیست، فرض کن خانه عملکرد معمولی (Typ) دارد.


# پر کردن ستون‌های عددی
#مقادیر صفر واسه این گذاشتم که این ستون‌ها مربوط به وجود یک بخش از خانه هستند. اگر گاراژ یا زیرزمین وجود نداشته باشد، منطقی است مقدارشان ۰ باشد.
#چرا LotFrontage و تعداد حمام‌ها را Median می‌گذاریم؟
#چون این‌ها ویژگی‌های واقعی خانه هستند و مقدار NaN لزوماً به معنی «ندارد» نیست. برای همین مقدار میانه (Median) انتخاب مناسب‌تری است.
#برای GarageYrBlt مقدار ۰ انتخاب مناسبی است
#چون وقتی همزمان ستون‌های GarageType، GarageArea و GarageCars هم صفر یا "No Garage" باشند
#مدل متوجه می‌شود که اصلاً گاراژی وجود نداشته است، نه اینکه سال ساخت آن نامشخص باشد.

test["GarageYrBlt"] = test["GarageYrBlt"].fillna(0)
test["MasVnrArea"] = test["MasVnrArea"].fillna(0)
test["BsmtFinSF1"] = test["BsmtFinSF1"].fillna(0)
test["TotalBsmtSF"] = test["TotalBsmtSF"].fillna(0)
test["BsmtUnfSF"] = test["BsmtUnfSF"].fillna(0)
test["BsmtFinSF2"] = test["BsmtFinSF2"].fillna(0)
test["GarageCars"] = test["GarageCars"].fillna(0)
test["GarageArea"] = test["GarageArea"].fillna(0)

#برای mode , median , ... # بهتره از دیتای آموزش استفاده بشه چون مدل با اون دیتا آموزش دیده
test["LotFrontage"] = test["LotFrontage"].fillna(train["LotFrontage"].median())
test["BsmtHalfBath"] = test["BsmtHalfBath"].fillna(train["BsmtHalfBath"].median())
test["BsmtFullBath"] = test["BsmtFullBath"].fillna(train["BsmtFullBath"].median())

#مهندسی فیچر

test["HouseAge"] = 2010 - test["YearBuilt"]

test["QualLivArea"] = test["OverallQual"] * test["GrLivArea"]

test["LotArea"] = np.log1p(test["LotArea"])

test["BsmtFinishedRatio"] = np.where(
    test["TotalBsmtSF"] > 0,#چون اگر TotalBsmtSF صفر باشد، از تقسیم بر صفر جلوگیری می‌شود.
    test["BsmtFinSF1"] / test["TotalBsmtSF"], 0)

test["QualCond"] = test["OverallQual"] * test["OverallCond"]

test["TotalBath"] = (
    test["FullBath"]
    + 0.5 * test["HalfBath"]
    + test["BsmtFullBath"]
    + 0.5 * test["BsmtHalfBath"])

test["TotalSF"] = test["TotalBsmtSF"] + test["GrLivArea"]

test["RemodAge"] = 2010 - test["YearRemodAdd"]

test["HasFireplace"] = (test["Fireplaces"] > 0).astype(int)

test["HasFence"] = (test["Fence"] != "No Fence").astype(int)

test["QualFireplace"] = test["OverallQual"] * test["Fireplaces"]
```
